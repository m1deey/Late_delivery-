```python
import duckdb

con = duckdb.connect()
con.register("orders", order)
con.register("customers", customers)
con.register("items", items)
con.register("product", product)
con.register("payment", payment)
con.register("review", review)
con.register("seller", seller)
```

```python
result = con.sql("""
SELECT
    o.order_id,
    o.customer_id,
    c.customer_unique_id,
    c.customer_city,
    c.customer_state
FROM orders o
JOIN customers c
    ON o.customer_id = c.customer_id
LIMIT 10
""")

result
```

```python
result = con.sql("""
SELECT order_status , COUNT(order_id) as number_of_ordera
FROM orders
GROUP BY  order_status
""")

result
```

```python
balance_check = con.sql("""
SELECT
    CASE WHEN order_delivered_customer_date > order_estimated_delivery_date
         THEN 1 ELSE 0 END AS is_late,
    COUNT(*) AS n_orders
FROM orders
WHERE order_status = 'delivered'
GROUP BY is_late
""")
balance_check
```

```python
result = con.sql("""
SELECT
    o.order_id,
    c.customer_state,
    s.seller_state,
    cat.product_category_name_english AS product_category,
    p.product_weight_g,
    p.product_length_cm,
    p.product_height_cm,
    p.product_width_cm,
    EXTRACT(month FROM o.order_purchase_timestamp) AS purchase_month,
    EXTRACT(dow FROM o.order_purchase_timestamp) AS purchase_dow,
    COUNT(i.order_item_id) AS n_items,
    SUM(i.price) AS total_price,
    SUM(i.freight_value) AS total_freight,
    AVG(pay.payment_installments) AS avg_installments,
    CASE WHEN o.order_delivered_customer_date > o.order_estimated_delivery_date
         THEN 1 ELSE 0 END AS is_late
FROM orders o
JOIN customers c ON o.customer_id = c.customer_id
JOIN items i ON o.order_id = i.order_id
JOIN seller s ON i.seller_id = s.seller_id
JOIN product p ON i.product_id = p.product_id
LEFT JOIN cat ON p.product_category_name = cat.product_category_name
JOIN payment pay ON o.order_id = pay.order_id
WHERE o.order_status = 'delivered'
GROUP BY o.order_id, c.customer_state, s.seller_state, cat.product_category_name_english,
         p.product_weight_g, p.product_length_cm, p.product_height_cm, p.product_width_cm,
         o.order_purchase_timestamp, o.order_delivered_customer_date, o.order_estimated_delivery_date
""")
df = result.df()
df.head(10)
```
